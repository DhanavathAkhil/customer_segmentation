```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, DBSCAN
from google.colab import drive

# Step 1: Mount Google Drive and load data
drive.mount('/content/drive')
file_path = '/content/drive/My Drive/ML/project/Mall_Customers.csv'
df = pd.read_csv(file_path)

# Display available columns for debugging
print("Columns in the dataset:", df.columns.tolist())

# Step 2: Data Cleaning
# Rename 'Genre' to 'Gender' if applicable (common in Mall_Customers dataset)
if 'Genre' in df.columns:
    df.rename(columns={'Genre': 'Gender'}, inplace=True)

# Removing duplicates
df.drop_duplicates(inplace=True)

# Handle inconsistent data in 'Gender' column (only if Gender exists)
if 'Gender' in df.columns:
    df['Gender'] = df['Gender'].str.strip().str.capitalize()
else:
    print("Warning: 'Gender' column not found. Skipping gender cleaning step.")

# Selecting features for clustering
X = df[['Annual Income (k$)', 'Spending Score (1-100)']].values

# K-Means Clustering with Elbow Method for comparison
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

# Plot the elbow graph
plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o')
plt.title('K-Means Elbow Point Graph')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()

# Apply KMeans for comparison with optimal clusters
optimal_clusters = 5  # You may change based on elbow graph
kmeans = KMeans(n_clusters=optimal_clusters, random_state=42)
df['KMeans_Cluster'] = kmeans.fit_predict(X)

# Step 3: DBSCAN Clustering
# Parameters eps and min_samples may need tuning
dbscan = DBSCAN(eps=7, min_samples=5)
df['DBSCAN_Cluster'] = dbscan.fit_predict(X)

# Step 4: Visualize the clusters

# KMeans Visualization
plt.figure(figsize=(8, 8))
for i in range(optimal_clusters):
    plt.scatter(X[df['KMeans_Cluster'] == i, 0], X[df['KMeans_Cluster'] == i, 1], s=50, label=f'KMeans Cluster {i+1}')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, c='black', marker='X', label='Centroids')
plt.title('KMeans Clustering')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()

# DBSCAN Visualization
plt.figure(figsize=(8, 8))
unique_labels = set(df['DBSCAN_Cluster'])
for label in unique_labels:
    if label == -1:
        color = 'red'
        label_name = 'Noise'
    else:
        color = plt.cm.jet(float(label) / (max(unique_labels) if max(unique_labels) > 0 else 1))
        label_name = f'Cluster {label + 1}'
    plt.scatter(X[df['DBSCAN_Cluster'] == label, 0], X[df['DBSCAN_Cluster'] == label, 1], s=50, color=color, label=label_name)
plt.title('DBSCAN Clustering')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()
```
